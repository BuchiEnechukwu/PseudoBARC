## Data Cleaning

I will standardise the columns to allow me to merge both frames together, and clean from one table. Then remove the obvious data gap in the df1, where the rows are empty.

#### Standardise the columns and merge both data

```python
# remove leading and trailing spaces from column names
df1.columns = df1.columns.astype(str).str.strip()
df2.columns = df2.columns.astype(str).str.strip()
```

```python
# rename df2 columns to ensure both match
# create dictionary of the columns names that need fixing
cols = {'PA': 'Pseudomonas',
        'visit': 'Visit',
        'FEV1%pred': 'FEV1_percent_pred',
        'Enterobactriaceae': 'Enterobacteriaceae',
        'M_catarrhalises': 'M_catarrhalis',
        'Dead 1=y, 0=no': 'Dead',
        'Total follow up': 'Total_follow_up',
        'Total Exs': 'Total_exacerbations',
        'TOTAL HOSP': 'Total_hospitalisation'
       }

# map to the datasets
df2 = df2.rename(columns=cols)
df1 = df1.rename(columns=cols)
```

```python
# reorder df2 columns to fit df1 
df2 = df2[df1.columns]

# confirm column order
print(df1.columns.equals(df2.columns))
```

```python
# combine the two datasets
# append df2 to df1
df = pd.concat([df1, df2], ignore_index=True)
```

```python
# standardise column names
# capitalise first letters
df.columns = [
    col[:1].upper() + col[1:] # ensure other letters do not become lower case
    for col in df.columns
]
```

### Handle duplicates and create new columns

```python
# check duplicate rows
df.duplicated().sum()
```

```python
# remove exact duplicate records
df = df.drop_duplicates().reset_index(drop=True)
```

While assessing the dataset on excel, I noticed a wide gap between records on the Europe data. I will remove them first, to allow me to encode and fix the real missing values in the columns.

```python
# find rows where every entry is missing
empty_rows = df[df.isna().all(axis=1)]
empty_rows.head()
```

```python
# drop empty rows
df = df.drop(empty_rows.index, axis=0)
```

#### Deal with variable types

```python
# understand the different types of responses in each variable
for col in df.columns:
    print('--------------------------------------------')
    unique_vals = df[col].unique().tolist()
    print(f'{col}:')
    print(f'{unique_vals}')
    print(f'Total Values: {len(unique_vals)}')
```

```python
# encode yes and no columns
# change the Yes to 1 and No to 0
yes_no_cols = [
    'Pseudomonas', 'Haemophilus', 'Enterobacteriaceae', 'S_aureus',
    'S_pneumoniae', 'M_catarrhalis', 'Ischaemic_heart_disease',
    'Stroke', 'Statin', 'ICS', 'COPD', 'Asthma', 'Diabetes', 'Liver_disease',
    'Osteoporosis', 'Depression', 'Neoplastic_disease', 'Chronic_renal_failure',]

val = {"No": 0, "Yes": 1}

# map the columns
for col in yes_no_cols:
    df[col] = df[col].map(val)
```

```python
# Sex
# change Male to 1 and Female to 0
df["Sex"] = df["Sex"].map({"Female":0, "Male": 1})
```

My inspection of the Days values showed that the decimal components represent plausible fractions of a day. For example, 357.0819165548044 represents approximately 357 days and 2 hours, while 408.8699415353718 represents approximately 408 days and 21 hours. Since Days represents elapsed time between visits and will be used in the longitudinal lung-function analysis, the original fractional values were retained. Rounding was avoided because it could reduce the precision of the recorded follow-up time without a clear analytical justification.

Continuous clinical measurements, including BMI, FVC, FEV1, FEV1_percent_pred, and FVC_percent_pred, will be retained at their original numerical precision to avoid loss of information. Rounding will be applied when presenting descriptive statistics, tables, and model results.

### Handle missing patient ID

4.88% of records are without PatientID.

```python
unknown_id = df[df["PatientID"].isna()]
len(unknown_id)
```

The records are significantly populated, which implies that they were not invalid records. Since the missing identifier prevents reliable linkage of repeated observations to individual patients, assigning these records to patients would require an unsupported assumption. However, let me check among patients with known IDs to see how often the combination of demographic/clinical characteristics corresponds to only one PatientID.

```python
# get known id records
known_id = df[~df["PatientID"].isna()]
```

```python
# candidate variables to test as a substitute identifier
match_cols = ["Age", "Height", "Sex", "Region", "BMI", "Ethnicity",
              "COPD", "Asthma", "Smoking"]

# compare match the variables
pct_unique = (
    known_id
    .dropna(subset=match_cols)           # one row per distinct (PatientID, profile) combination
    .groupby(match_cols, observed=True)["PatientID"]           # group by the demographic/clinical profile
    .nunique()                # count how many distinct PatientIDs share each profile
    .eq(1)                # True where a profile maps to exactly one patient (unique)
    .mean()              # fraction of profiles that are uniquely identifying
)

pct_unique
```

100% of demographic/clinical profiles map to exactly one known patient. This suggests that the variables are highly discriminating. But let me try the non-known patient ID records to see how they match with those with known.

```python
# extract records with patientID records
lookup = (
    known_id[["PatientID"] + match_cols]  # keep PatientID + match_cols
    .dropna(subset=match_cols)       # drop rows with incomplete profiles (can't use as lookup key)
    .drop_duplicates(subset=match_cols)       # keep one row per profile
    .rename(columns={"PatientID": "matched_PatientID"})       # rename to avoid clashing with df's PatientID on merge
)
```

```python
lookup
```

```python
# attach a matched_PatientID where the profile matches a known patient
unknown_match = (unknown_id.merge(lookup, on=match_cols, how="left"))
```

```python
# count match vs unmatched
unknown_match["matched_PatientID"].notna().value_counts()  
```

Exact matching of the 1,330 records with missing PatientID against those with matching profiles identified 654 that could be matched, while 676 records remained unmatched. I will attach the lookup to the df, then keep the linked data in a separate saved file, and use it as a sensitivty analysis to confirm robustness. Meanwhile, the decision on the use will be confirmation from Prof Chalmers

Q4: Can records with missing PatientID be linked to existing patients using exact matches on the static patient characteristics?

```python
# attach the matched iD
df = df.merge(lookup, on=match_cols, how="left")
```

```python
# track source of ID for ease of identification
def id_source(row):
    """
    Assigns original if the record had PatientID, then linked if part of the link (part of the 654)
    and unresolved if it cannot be matched (part of the 676)
    """
    if pd.notna(row["PatientID"]):
        return "original"
        
    elif pd.notna(row["matched_PatientID"]):
        return "linked"
        
    else:
        return "not linked"

# apply to the dataset
df["ID_source"] = df.apply(id_source, axis=1)
```

```python
df["ID_source"].value_counts()
```

```python
df_primary = df[df["ID_source"] == "original"].copy()
df_primary = df_primary.drop('matched_PatientID', axis=1)
```

```python
# filter for sensitivity
df_sensitivity = df[
    df["ID_source"].isin(["original", "linked"])
].copy()
```

```python
# fill the missing PatientID with the matched ID
df_sensitivity["PatientID"] = (
    df_sensitivity["PatientID"].astype("string")
    .fillna(df_sensitivity["matched_PatientID"].astype("string"))
    .astype("category")
)
```

```python
# check destribution of the visits
visits = df_primary.groupby("Visit", observed=True)["Days"].agg(
    ["count", "min", "median", "max"])

visits
```

```python
# V1 is apparent 0
# assign V1 with 0
df_primary.loc[
    (df_primary["Visit"] == "V1") & df_primary["Days"].isna(),
    "Days"] = 0
```

```python
# fill where 
def assign_visit(day):
    """Assign missing Visit from the observed Days distribution."""
    
    if pd.isna(day):
        return np.nan
    elif day == 0:
        return "V1"
    elif 243.005416 <= day <= 477.434860:
        return "V5"
    elif 587.588575 <= day <= 873.259868:
        return "V6"
    else:
        return np.nan
```

```python
#  map the function
df_primary["Visit"] = df_primary["Visit"].fillna(
    df_primary["Days"].apply(assign_visit)
)
```

```python
# confirm the na in both columns
pd.crosstab(
    df_primary["Visit"].isna(),
    df_primary["Days"].isna()
)
```

```python
# find teh one record with the NA for visit
df_primary.loc[
    df_primary["Visit"].isna() & df_primary["Days"].notna(),
    ["PatientID", "Visit", "Days"]]
```

The days fall outside the range, so I will remove it.

```python
# drop the row
df_primary = df_primary.drop(
    df_primary[
        df_primary["Visit"].isna() &
        df_primary["Days"].notna()
    ].index
)
```

# Question 1: adjusted differences in lung function

This question depends on the mulple visit by patients. I will need an algorithm that can account for that. Multiple linear regression would be appropriate if the analysis were restricted to a single observation per patient, such as the baseline visit, given that the dataset contains repeated lung-function measurements across multiple visits for some individuals. Using ordinary multiple linear regression on all observations would not account for the correlation between measurements from the same patient. A linear mixed-effects model, with PatientID included as a random effect, would allow all available visits to be used while accounting for within-patient repeated measurements.

https://www.statsmodels.org/stable/generated/statsmodels.formula.api.mixedlm.html#

```python
# create the data
model_df = df_primary[
    ["PatientID", "FEV1_percent_pred", "comorbidity_group",
     "Age", "Sex", "BMI", "Smoking", "Region", "Ethnicity",
     "Years"]
].dropna()
```

Equation: \
FEV1% predicted = COPD/asthma group + age + sex + BMI + smoking + region + ethnicity + time

```python
model_formula = (
    "FEV1_percent_pred ~ "
    "C(comorbidity_group, Treatment(reference='Neither')) + "
    "Age + C(Sex) + BMI + C(Smoking) + "
    "C(Region) + C(Ethnicity) + Years"
)
```

```python
model = smf.mixedlm(
    model_formula,
    data=model_df,
    groups=model_df["PatientID"]
)
```

```python
#  fit the model
result = model.fit()

print(result.summary())
```

```python
# fitted values and residuals
fitted = result.fittedvalues
residuals = result.resid

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Residuals vs fitted
sns.scatterplot(
    x=fitted,
    y=residuals,
    alpha=0.3,
    ax=axes[0]
)

axes[0].axhline(0, linestyle="--")
axes[0].set_title("Residuals vs Fitted Values")
axes[0].set_xlabel("Fitted FEV1% predicted")
axes[0].set_ylabel("Residuals")

# Q-Q plot
sm.qqplot(
    residuals,
    line="45",
    ax=axes[1]
)

axes[1].set_title("Q-Q Plot of Residuals")

plt.tight_layout()
plt.show()
```
